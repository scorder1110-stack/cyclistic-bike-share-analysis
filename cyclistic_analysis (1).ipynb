```python
station_trips = analysis_trips[
    analysis_trips["start_station_name"].notna()
].copy()

print("Total analysis rides:", len(analysis_trips))
print("Rides with start station:", len(station_trips))
print("Rides without start station:", len(analysis_trips) - len(station_trips))
```

```python
top_start_stations = (
    station_trips
    .groupby(["member_casual", "start_station_name"])
    .size()
    .reset_index(name="ride_count")
)

top_casual_stations = (
    top_start_stations[
        top_start_stations["member_casual"] == "casual"
    ]
    .sort_values("ride_count", ascending=False)
    .head(10)
)

top_member_stations = (
    top_start_stations[
        top_start_stations["member_casual"] == "member"
    ]
    .sort_values("ride_count", ascending=False)
    .head(10)
)

print("Top 10 casual starting stations:")
print(top_casual_stations.to_string(index=False))

print("\nTop 10 member starting stations:")
print(top_member_stations.to_string(index=False))
```

```python
station_totals = (
    station_trips
    .groupby(["member_casual", "start_station_name"])
    .size()
    .reset_index(name="ride_count")
)

for member_type in ["casual", "member"]:
    
    group = station_totals[
        station_totals["member_casual"] == member_type
    ].sort_values("ride_count", ascending=False)
    
    total = group["ride_count"].sum()
    
    top_10 = group.head(10)["ride_count"].sum()
    top_25 = group.head(25)["ride_count"].sum()
    top_50 = group.head(50)["ride_count"].sum()
    
    print(f"\n{member_type.upper()}")
    print(f"Total rides with known start station: {total:,}")
    print(f"Top 10 stations: {top_10:,} ({top_10 / total * 100:.2f}%)")
    print(f"Top 25 stations: {top_25:,} ({top_25 / total * 100:.2f}%)")
    print(f"Top 50 stations: {top_50:,} ({top_50 / total * 100:.2f}%)")
```

```python
station_mix = (
    station_trips
    .groupby(["start_station_name", "member_casual"])
    .size()
    .unstack(fill_value=0)
)

station_mix["total_rides"] = station_mix.sum(axis=1)

station_mix["casual_pct"] = (
    station_mix["casual"] /
    station_mix["total_rides"] * 100
)

station_opportunities = (
    station_mix[
        station_mix["total_rides"] >= 1000
    ]
    .sort_values(
        ["casual_pct", "total_rides"],
        ascending=[False, False]
    )
)

print("Top stations by casual rider percentage:")
print(
    station_opportunities[
        ["casual", "member", "total_rides", "casual_pct"]
    ].head(20).round(2)
)
```

```python
casual_station_daytype = (
    station_trips[station_trips["member_casual"] == "casual"]
    .groupby(["start_station_name", "day_type"])
    .size()
    .unstack(fill_value=0)
)

casual_station_daytype["total_casual_rides"] = (
    casual_station_daytype["Weekday"] +
    casual_station_daytype["Weekend"]
)

casual_station_daytype["weekday_pct"] = (
    casual_station_daytype["Weekday"] /
    casual_station_daytype["total_casual_rides"] * 100
)

weekday_casual_stations = (
    casual_station_daytype[
        casual_station_daytype["total_casual_rides"] >= 1000
    ]
    .sort_values(
        ["weekday_pct", "total_casual_rides"],
        ascending=[False, False]
    )
)

print(
    weekday_casual_stations[
        ["Weekday", "Weekend", "total_casual_rides", "weekday_pct"]
    ].head(20).round(2)
)
```

```python
casual_commute_window = station_trips[
    (station_trips["member_casual"] == "casual") &
    (
        station_trips["hour"].isin([7, 8, 9, 16, 17, 18])
    )
].copy()

station_commute_pattern = (
    casual_commute_window
    .groupby("start_station_name")
    .size()
    .reset_index(name="casual_peak_hour_rides")
    .sort_values("casual_peak_hour_rides", ascending=False)
)

print("Top stations for casual rides during member peak hours:")
print(
    station_commute_pattern.head(20).to_string(index=False)
)
```

```python
# Total casual rides by station
casual_station_total = (
    station_trips[station_trips["member_casual"] == "casual"]
    .groupby("start_station_name")
    .size()
    .reset_index(name="total_casual_rides")
)

# Merge total casual rides with peak-hour casual rides
casual_peak_analysis = casual_station_total.merge(
    station_commute_pattern,
    on="start_station_name",
    how="left"
)

# Stations with no rides in the peak window get zero
casual_peak_analysis["casual_peak_hour_rides"] = (
    casual_peak_analysis["casual_peak_hour_rides"]
    .fillna(0)
)

# Calculate percentage of casual rides occurring during member peak hours
casual_peak_analysis["peak_hour_pct"] = (
    casual_peak_analysis["casual_peak_hour_rides"]
    / casual_peak_analysis["total_casual_rides"]
    * 100
)

# Require at least 1,000 casual rides to avoid tiny samples
casual_peak_analysis = casual_peak_analysis[
    casual_peak_analysis["total_casual_rides"] >= 1000
].copy()

# Sort by percentage
casual_peak_analysis = casual_peak_analysis.sort_values(
    "peak_hour_pct",
    ascending=False
)

print("Top stations by percentage of casual rides during member peak hours:")
print(
    casual_peak_analysis[
        [
            "start_station_name",
            "total_casual_rides",
            "casual_peak_hour_rides",
            "peak_hour_pct"
        ]
    ].head(25).to_string(index=False)
)
```

```python
# Focus on casual riders at stations with meaningful casual volume
casual_station_day_duration = (
    station_trips[
        station_trips["member_casual"] == "casual"
    ]
    .groupby(
        ["start_station_name", "day_type"]
    )["ride_length"]
    .agg(
        rides="count",
        average_minutes="mean",
        median_minutes="median"
    )
    .reset_index()
)

# Keep stations with at least 1,000 total casual rides
casual_station_volume = (
    station_trips[
        station_trips["member_casual"] == "casual"
    ]
    .groupby("start_station_name")
    .size()
    .reset_index(name="total_casual_rides")
)

casual_station_day_duration = casual_station_day_duration.merge(
    casual_station_volume,
    on="start_station_name",
    how="left"
)

casual_station_day_duration = casual_station_day_duration[
    casual_station_day_duration["total_casual_rides"] >= 1000
].copy()

# Display stations with weekday and weekend observations
print(
    casual_station_day_duration
    .sort_values(
        ["start_station_name", "day_type"]
    )
    .head(50)
    .to_string(index=False)
)
```

```python
# Create one row per station with weekday and weekend casual duration
casual_station_duration_pivot = (
    casual_station_day_duration
    .pivot(
        index="start_station_name",
        columns="day_type",
        values=["rides", "average_minutes", "median_minutes"]
    )
)

# Flatten column names
casual_station_duration_pivot.columns = [
    f"{metric}_{day_type.lower()}"
    for metric, day_type in casual_station_duration_pivot.columns
]

casual_station_duration_pivot = (
    casual_station_duration_pivot
    .reset_index()
)

# Calculate weekend vs. weekday differences
casual_station_duration_pivot["avg_weekend_minus_weekday"] = (
    casual_station_duration_pivot["average_minutes_weekend"]
    - casual_station_duration_pivot["average_minutes_weekday"]
)

casual_station_duration_pivot["median_weekend_minus_weekday"] = (
    casual_station_duration_pivot["median_minutes_weekend"]
    - casual_station_duration_pivot["median_minutes_weekday"]
)

# Add total casual rides
casual_station_duration_pivot = casual_station_duration_pivot.merge(
    casual_station_volume,
    on="start_station_name",
    how="left"
)

# Show stations where weekends are most different from weekdays
print(
    casual_station_duration_pivot
    .sort_values("avg_weekend_minus_weekday", ascending=False)
    .head(30)
    .to_string(index=False)
)
```

```python
# Create casual_peak_station from casual_peak_analysis for downstream compatibility
casual_peak_station = casual_peak_analysis[
    [
        "start_station_name",
        "casual_peak_hour_rides",
        "peak_hour_pct"
    ]
].rename(
    columns={
        "casual_peak_hour_rides": "casual_peak_rides",
        "peak_hour_pct": "casual_peak_pct"
    }
)

# Build station-level casual behavior profile
casual_daytype_volume = (
    station_trips[
        station_trips["member_casual"] == "casual"
    ]
    .groupby(["start_station_name", "day_type"])
    .size()
    .unstack(fill_value=0)
    .reset_index()
)

casual_daytype_volume.columns.name = None

casual_daytype_volume = casual_daytype_volume.rename(
    columns={
        "Weekday": "casual_weekday_rides",
        "Weekend": "casual_weekend_rides"
    }
)

# Station membership mix
station_membership = (
    station_trips
    .groupby(["start_station_name", "member_casual"])
    .size()
    .unstack(fill_value=0)
    .reset_index()
)

station_membership.columns.name = None

if "casual" not in station_membership.columns:
    station_membership["casual"] = 0

if "member" not in station_membership.columns:
    station_membership["member"] = 0

station_membership["station_total_rides"] = (
    station_membership["casual"] +
    station_membership["member"]
)

station_membership["casual_pct"] = (
    station_membership["casual"] /
    station_membership["station_total_rides"] *
    100
)

station_membership = station_membership.rename(
    columns={
        "casual": "casual_rides",
        "member": "member_rides"
    }
)

# Merge duration and peak station information
station_opportunity = (
    station_membership[
        [
            "start_station_name",
            "casual_rides",
            "member_rides",
            "station_total_rides",
            "casual_pct"
        ]
    ]
    .merge(
        casual_daytype_volume,
        on="start_station_name",
        how="left"
    )
    .merge(
        casual_station_duration_pivot[
            [
                "start_station_name",
                "average_minutes_weekday",
                "average_minutes_weekend",
                "median_minutes_weekday",
                "median_minutes_weekend"
            ]
        ],
        on="start_station_name",
        how="left"
    )
    .merge(
        casual_peak_station[
            [
                "start_station_name",
                "casual_peak_rides",
                "casual_peak_pct"
            ]
        ],
        on="start_station_name",
        how="left"
    )
)

# Fill missing peak values
station_opportunity["casual_peak_rides"] = station_opportunity["casual_peak_rides"].fillna(0)
station_opportunity["casual_peak_pct"] = station_opportunity["casual_peak_pct"].fillna(0)

# Keep stations with at least 1,000 casual rides
station_opportunity = station_opportunity[
    station_opportunity["casual_rides"] >= 1000
].copy()

# Calculate weekday share of casual rides
station_opportunity["casual_weekday_pct"] = (
    station_opportunity["casual_weekday_rides"] /
    station_opportunity["casual_rides"] *
    100
)

# Sort by casual volume
print(
    station_opportunity
    .sort_values("casual_rides", ascending=False)
    .head(40)
    .to_string(index=False)
)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Create duration distribution by rider type
duration_summary = (
    analysis_trips
    .groupby(
        ["member_casual", "duration_group"],
        observed=False
    )
    .size()
    .reset_index(name="rides")
)

# Calculate percentage within each rider type
duration_summary["percentage"] = (
    duration_summary["rides"]
    / duration_summary.groupby(
        "member_casual",
        observed=False
    )["rides"].transform("sum")
    * 100
)

# Set desired duration order
duration_order = [
    "Under 10 min",
    "10–20 min",
    "20–30 min",
    "30–60 min",
    "60+ min"
]

# Pivot for plotting
duration_plot = (
    duration_summary
    .pivot(
        index="member_casual",
        columns="duration_group",
        values="percentage"
    )
    .reindex(columns=duration_order)
    .fillna(0)
)

# Put casual first, member second
duration_plot = duration_plot.reindex(["casual", "member"])

# Brighter blue/orange palette to match the other charts
duration_colors = [
    "#4E79A7",  # Under 10 min
    "#76A5C0",  # 10–20 min
    "#59A14F",  # 20–30 min
    "#F28E2B",  # 30–60 min
    "#E15759"   # 60+ min
]

# Create chart
fig, ax = plt.subplots(figsize=(10, 5.5))

duration_plot.plot(
    kind="barh",
    stacked=True,
    ax=ax,
    color=duration_colors
)

ax.set_title(
    "Ride Duration Distribution by Rider Type",
    fontsize=16,
    fontweight="bold",
    loc="left"
)

ax.set_xlabel("Share of rides (%)")
ax.set_ylabel("")

ax.set_xlim(0, 100)

ax.legend(
    title="Ride duration",
    bbox_to_anchor=(1.02, 1),
    loc="upper left",
    frameon=False
)

ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)

plt.tight_layout()

plt.savefig(
    "images/ride_duration_distribution.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

```python
member_station_behavior = (
    analysis_trips[
        (analysis_trips["member_casual"] == "member") &
        (analysis_trips["start_station_name"].isin(
            conversion_candidates["start_station_name"]
        ))
    ]
    .groupby("start_station_name")
    .agg(
        member_rides=("ride_id", "count"),
        member_average_minutes=("ride_length", "mean"),
        member_weekday_rides=(
            "day_type",
            lambda x: (x == "Weekday").sum()
        )
    )
)

member_station_behavior["member_weekday_pct"] = (
    member_station_behavior["member_weekday_rides"]
    / member_station_behavior["member_rides"]
    * 100
)

print(member_station_behavior.head(10))
```

```python
member_peak_hour = (
    analysis_trips[
        (analysis_trips["member_casual"] == "member") &
        (analysis_trips["start_station_name"].isin(
            conversion_candidates["start_station_name"]
        ))
    ]
    .groupby("start_station_name")
    .agg(
        member_peak_hour_rides=(
            "hour",
            lambda x: x.isin([7, 8, 9, 16, 17, 18]).sum()
        )
    )
)

member_station_behavior = member_station_behavior.join(
    member_peak_hour
)

member_station_behavior["member_peak_hour_pct"] = (
    member_station_behavior["member_peak_hour_rides"]
    / member_station_behavior["member_rides"]
    * 100
)

print(member_station_behavior.head(10))
```

```python
conversion_comparison = (
    conversion_candidates[
        [
            "start_station_name",
            "casual_rides",
            "average_minutes_weekday",
            "casual_weekday_pct",
            "peak_hour_pct"
        ]
    ]
    .set_index("start_station_name")
    .join(
        member_station_behavior[
            [
                "member_rides",
                "member_average_minutes",
                "member_weekday_pct",
                "member_peak_hour_pct"
            ]
        ]
    )
)

print(
    conversion_comparison
    .head(10)
    .round(1)
    .to_string()
)
```

```python
import matplotlib.pyplot as plt

comparison_chart = conversion_comparison.head(10).sort_values(
    "casual_weekday_pct"
)

plt.figure(figsize=(12, 7))

x = range(len(comparison_chart))
width = 0.4

# Casual = muted blue
plt.bar(
    [i - width / 2 for i in x],
    comparison_chart["casual_weekday_pct"],
    width=width,
    label="Casual",
    color=casual_color
)

# Member = muted orange
plt.bar(
    [i + width / 2 for i in x],
    comparison_chart["member_weekday_pct"],
    width=width,
    label="Member",
    color=member_color
)

plt.xticks(
    x,
    comparison_chart.index,
    rotation=45,
    ha="right"
)

plt.ylabel("Weekday Rides (%)")
plt.xlabel("Starting Station")

plt.title(
    "Weekday Usage: Routine-Like Casual Riders vs. Members",
    fontsize=16,
    fontweight="bold"
)

plt.legend()

plt.tight_layout()

plt.savefig(
    "images/weekday_casual_vs_member_conversion.svg",
    format="svg",
    bbox_inches="tight"
)

plt.show()
```
